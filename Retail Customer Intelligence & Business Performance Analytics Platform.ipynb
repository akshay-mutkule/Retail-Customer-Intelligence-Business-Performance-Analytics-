```python
customer_summary = (
    df.groupby("Customer ID")
      .agg(
          Customer_Name=("Customer Name", "first"),
          Total_Orders=("Order ID", "nunique"),
          Total_Sales=("Sales", "sum"),
          Total_Profit=("Profit", "sum"),
          Total_Quantity=("Quantity", "sum"),
          Avg_Discount=("Discount", "mean"),
          First_Order=("Order Date", "min"),
          Last_Order=("Order Date", "max")
      )
      .reset_index()
)

customer_summary.head()
```

```python
#Customer Lifetime
customer_summary["Customer_Lifetime_Days"] = (
    customer_summary["Last_Order"] -
    customer_summary["First_Order"]
).dt.days
```

```python
#Average Order Value (AOV)
customer_summary["Average_Order_Value"] = (
    customer_summary["Total_Sales"] /
    customer_summary["Total_Orders"]
)
```

```python
#Customer Lifetime Value (CLV)
customer_summary["CLV"] = (
    customer_summary["Average_Order_Value"] *
    customer_summary["Total_Orders"]
)
```

```python
#Purchase Frequency
customer_summary["Purchase_Frequency"] = (
    customer_summary["Total_Orders"] /
    (customer_summary["Customer_Lifetime_Days"] + 1)
)
```

```python
#Top 10 Customers
top_customers = (
    customer_summary
    .sort_values("Total_Sales", ascending=False)
    .head(10)
)

top_customers
```

```python
#Customer Ranking
customer_summary["Customer_Rank"] = (
    customer_summary["Total_Sales"]
    .rank(method="dense", ascending=False)
)
```

```python
#VIP Customers
customer_summary["Customer_Type"] = pd.qcut(
    customer_summary["Total_Sales"],
    q=4,
    labels=[
        "Bronze",
        "Silver",
        "Gold",
        "Platinum"
    ]
)
```

```python
#Customer Segment Summary
customer_summary.groupby("Customer_Type").agg(
    Customers=("Customer ID", "count"),
    Sales=("Total_Sales", "sum"),
    Avg_CLV=("CLV", "mean")
)
```

# Phase 4: Customer Intelligence – RFM Segmentation & Behavioral Analysis

## Objective

RFM (Recency, Frequency, Monetary) Analysis helps identify high-value customers based on their purchasing behavior.

### RFM Metrics

- **Recency (R):** Days since the customer's last purchase.
- **Frequency (F):** Number of unique orders placed.
- **Monetary (M):** Total sales generated by the customer.

### Business Objectives

- Identify Champions and Loyal Customers.
- Find At-Risk Customers.
- Improve Customer Retention.
- Support Marketing Campaigns.
- Increase Customer Lifetime Value.

```python
# ============================================
# Step 4.1 : Define Analysis Date
# ============================================

# Analysis date is one day after the latest order date

analysis_date = df["Order Date"].max() + pd.Timedelta(days=1)

print("Analysis Date :", analysis_date)
```

```python
# ============================================
# Step 4.2 : Create RFM Table
# ============================================

rfm = (
    df.groupby("Customer ID")
      .agg({
          "Order Date": lambda x: (analysis_date - x.max()).days,
          "Order ID": "nunique",
          "Sales": "sum"
      })
      .reset_index()
)

# Rename columns

rfm.columns = [
    "Customer ID",
    "Recency",
    "Frequency",
    "Monetary"
]

rfm.head()
```

```python
# ============================================
# Step 4.3 : Explore RFM Data
# ============================================

rfm.describe()
```

```python
# ============================================
# Step 4.4 : Create Recency Score
# ============================================

# Lower recency is better

rfm["R_Score"] = pd.qcut(
    rfm["Recency"],
    q=5,
    labels=[5,4,3,2,1]
)

rfm.head()
```

```python
# ============================================
# Step 4.5 : Create Frequency Score
# ============================================

# Higher frequency is better

rfm["F_Score"] = pd.qcut(
    rfm["Frequency"].rank(method="first"),
    q=5,
    labels=[1,2,3,4,5]
)

rfm.head()
```

```python
# ============================================
# Step 4.6 : Create Monetary Score
# ============================================

# Higher monetary value is better

rfm["M_Score"] = pd.qcut(
    rfm["Monetary"],
    q=5,
    labels=[1,2,3,4,5]
)

rfm.head()
```

```python
# ============================================
# Step 4.7 : Generate RFM Score
# ============================================

rfm["RFM Score"] = (
    rfm["R_Score"].astype(str)
    + rfm["F_Score"].astype(str)
    + rfm["M_Score"].astype(str)
)

rfm.head()
```

```python
# ============================================
# Step 4.8 : Calculate Total RFM Score
# ============================================

rfm["RFM Total"] = (
    rfm["R_Score"].astype(int)
    + rfm["F_Score"].astype(int)
    + rfm["M_Score"].astype(int)
)

rfm.head()
```

```python
# ============================================
# Step 4.9 : Create Customer Segments
# ============================================

def customer_segment(score):

    if score >= 13:
        return "Champions"

    elif score >= 11:
        return "Loyal Customers"

    elif score >= 9:
        return "Potential Loyalists"

    elif score >= 7:
        return "Promising"

    elif score >= 5:
        return "At Risk"

    else:
        return "Hibernating"
```

```python
# ============================================
# Step 4.10 : Apply Customer Segmentation
# ============================================

rfm["Customer Segment"] = rfm["RFM Total"].apply(customer_segment)

rfm.head()
```

```python
# ============================================
# Step 4.11 : Segment Summary
# ============================================

segment_summary = (
    rfm.groupby("Customer Segment")
       .agg(
           Customers=("Customer ID","count"),
           Avg_Recency=("Recency","mean"),
           Avg_Frequency=("Frequency","mean"),
           Avg_Monetary=("Monetary","mean")
       )
       .sort_values("Avg_Monetary", ascending=False)
)

segment_summary
```

```python
# ============================================
# Step 4.12 : Customer Segment Distribution
# ============================================

rfm["Customer Segment"].value_counts()
```

```python
# ============================================
# Step 4.13 : Visualize Customer Segments
# ============================================

rfm["Customer Segment"].value_counts().plot(
    kind="bar",
    figsize=(8,5)
)

plt.title("Customer Segment Distribution")

plt.xlabel("Customer Segment")

plt.ylabel("Number of Customers")

plt.xticks(rotation=20)

plt.show()
```

```python
# ============================================
# Step 4.14 : Top 20 Customers
# ============================================

top_customers = (
    rfm.sort_values(
        "Monetary",
        ascending=False
    )
    .head(20)
)

top_customers
```

```python
# ============================================
# Step 4.15 : Export RFM Analysis
# ============================================

rfm.to_csv(
    "customer_rfm_analysis.csv",
    index=False
)

print("Customer RFM Analysis Exported Successfully.")
```

```python
# ==========================================================
# 📊 Phase 6 – Predictive Analytics & Machine Learning
# Section 6.1 – Customer Segmentation (K-Means Clustering)
# Step 6.1.1 – Import Required Libraries
# ==========================================================

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

import warnings
warnings.filterwarnings("ignore")
```

```python
# ==========================================================
# Step 6.1.2 – Select Customer Features
# ==========================================================

cluster_features = customer_summary[
    [
        'Total_Sales',
        'Total_Orders',
        'Total_Profit',
        'Total_Quantity',
        'Average_Order_Value',
        'Purchase_Frequency',
        'CLV'
    ]
]

cluster_features.head()
```

```python
print(customer_summary.shape)
customer_summary.head()
```

```python
# ==========================================================
# Step 6.1.3 – Check Missing Values
# ==========================================================

print(cluster_features.isnull().sum())
```

```python
# ==========================================================
# Step 6.1.4 – Standardize Features
# ==========================================================

scaler = StandardScaler()

scaled_features = scaler.fit_transform(cluster_features)

scaled_features[:5]
```

```python
# ==========================================================
# Step 6.1.5 – Verify Scaled Dataset
# ==========================================================

print("Scaled Dataset Shape:", scaled_features.shape)
```

```python
# ==========================================================
# Step 6.1.6 – Elbow Method
# ==========================================================

wcss = []

for k in range(2, 11):

    kmeans = KMeans(
        n_clusters=k,
        init='k-means++',
        random_state=42,
        n_init=10
    )

    kmeans.fit(scaled_features)

    wcss.append(kmeans.inertia_)


# ==========================================================
# Plot Elbow Curve
# ==========================================================

plt.figure(figsize=(10, 6))

plt.plot(
    range(2, 11),
    wcss,
    marker='o',
    linewidth=3,
    markersize=8
)

plt.title("Elbow Method for Optimal K", fontsize=16)
plt.xlabel("Number of Clusters (K)")
plt.ylabel("WCSS")
plt.grid(True)

plt.show()
```

```python
# ==========================================================
# Step 6.1.7 – Silhouette Score
# ==========================================================

silhouette_scores = []

for k in range(2,11):

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = model.fit_predict(scaled_features)

    score = silhouette_score(
        scaled_features,
        labels
    )

    silhouette_scores.append(score)

# Display Scores
for k, score in zip(range(2,11), silhouette_scores):
    print(f"K = {k}  --->  Silhouette Score = {score:.4f}")
```

```python
# ==========================================================
# Step 6.1.8 – Plot Silhouette Scores
# ==========================================================

plt.figure(figsize=(10,6))

plt.plot(
    range(2,11),
    silhouette_scores,
    marker='o',
    linewidth=3,
    markersize=8
)

plt.title("Silhouette Score for Different K", fontsize=16)
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Silhouette Score")
plt.grid(True)

plt.show()
```

```python
# ==========================================================
# Step 6.1.8 – Plot Silhouette Scores
# ==========================================================

plt.figure(figsize=(10,6))

plt.plot(
    range(2,11),
    silhouette_scores,
    marker='o',
    linewidth=3,
    markersize=8
)

plt.title("Silhouette Score for Different K", fontsize=16)
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Silhouette Score")
plt.grid(True)

plt.show()
```

```python
# ==========================================================
# Step 6.1.9 – Identify Best K Using Silhouette Score
# ==========================================================

best_k = range(2, 11)[np.argmax(silhouette_scores)]

best_score = max(silhouette_scores)

print("Best K:", best_k)
print("Best Silhouette Score:", round(best_score, 4))
```

```python
# ==========================================================
# Step 6.5.5 – Logistic Regression Evaluation
# ==========================================================

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)

logistic_accuracy = accuracy_score(
    y_test_cls,
    logistic_predictions
)

logistic_precision = precision_score(
    y_test_cls,
    logistic_predictions
)

logistic_recall = recall_score(
    y_test_cls,
    logistic_predictions
)

logistic_f1 = f1_score(
    y_test_cls,
    logistic_predictions
)

print("Logistic Regression Results")
print("--------------------------------")
print("Accuracy :", round(logistic_accuracy, 4))
print("Precision:", round(logistic_precision, 4))
print("Recall   :", round(logistic_recall, 4))
print("F1 Score :", round(logistic_f1, 4))

print("\nClassification Report:")
print(
    classification_report(
        y_test_cls,
        logistic_predictions
    )
)
```

```python
# ==========================================================
# Step 6.5.6 – Random Forest Classifier
# ==========================================================

from sklearn.ensemble import RandomForestClassifier

rf_classifier = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    n_jobs=-1
)

rf_classifier.fit(
    X_train_cls,
    y_train_cls
)

rf_class_predictions = rf_classifier.predict(
    X_test_cls
)

print("Random Forest Classifier trained successfully.")
```
